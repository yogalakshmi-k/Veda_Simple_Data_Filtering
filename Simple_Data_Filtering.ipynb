```python

# Veda Technology - AI & ML Internship
# Day 8: Simple Data Filtering

# Create student records using a list of dictionaries
students = [
    {"name": "Arun", "age": 21, "marks": 85, "department": "CSE"},
    {"name": "Priya", "age": 22, "marks": 92, "department": "IT"},
    {"name": "Kavin", "age": 20, "marks": 67, "department": "CSE"},
    {"name": "Divya", "age": 21, "marks": 78, "department": "AI & ML"},
    {"name": "Rahul", "age": 23, "marks": 55, "department": "IT"},
    {"name": "Meena", "age": 20, "marks": 88, "department": "AI & ML"}
]

print("ALL STUDENT RECORDS")
for student in students:
    print(student)

# Filter 1: Students scoring 80 or above
print("\nFILTER 1: Students with marks >= 80")
high_marks = [s for s in students if s["marks"] >= 80]
for student in high_marks:
    print(student)

# Filter 2: Students from the CSE department
print("\nFILTER 2: Students in CSE Department")
cse_students = [s for s in students if s["department"] == "CSE"]
for student in cse_students:
    print(student)

# Filter 3: Students aged 21 or below
print("\nFILTER 3: Students aged <= 21")
young_students = [s for s in students if s["age"] <= 21]
for student in young_students:
    print(student)

# Filter 4: AI & ML students scoring 80 or above
print("\nFILTER 4: AI & ML students with marks >= 80")
ai_high_marks = [
    s for s in students
    if s["department"] == "AI & ML" and s["marks"] >= 80
]
for student in ai_high_marks:
    print(student)
```
